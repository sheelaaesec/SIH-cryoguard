```python
import numpy as np
import matplotlib.pyplot as plt

# =====================================================================
# 1. TIME STEP CONFIGURATION (24-Hour Cycle)
# =====================================================================
hours = np.linspace(0, 24, 1440) # 1440 minutes in a day
dt = 60.0 # Time step in seconds (1 minute steps)

# =====================================================================
# 2. ENVIRONMENTAL PARAMETERS (Ladakh Winter Profile)
# =====================================================================
# Ambient temperature curve: Low of -22°C at 4 AM, High of -2°C at 2 PM
T_ambient = -12 + 10 * np.sin((hours - 8) * np.pi / 12)

# Solar Radiation Profile: Peaks at 850 W/m² during midday, 0 W/m² at night
Q_solar_flux = np.maximum(0, 850 * np.sin((hours - 6) * np.pi / 11))

# =====================================================================
# 3. BASELINE VS OPTIMIZED PASSED SHELTER PARAMETERS
# =====================================================================
A_window = 3.0  # Window aperture area (m²)
Solar_SHGC = 0.7 # Solar Heat Gain Coefficient (efficiency of glazing)

# Shelter Variant A: Unoptimized Base Shelter (Low insulation, low mass)
R_base = 0.05    # Thermal resistance (K/W) - Thin walls, high leaks
C_base = 150000  # Thermal capacitance (J/K) - Tin/Wood sheets, no storage

# Shelter Variant B: Optimized Passive Design (High insulation, high thermal mass)
R_opt = 0.45     # Thermal resistance (K/W) - Multi-material insulation stack
C_opt = 1800000  # Thermal capacitance (J/K) - Phase change mass / mud bricks

# =====================================================================
# 4. TRANSIENT SIMULATION LOOP (Finite Difference Method)
# =====================================================================
T_inside_base = np.zeros_like(hours)
T_inside_opt = np.zeros_like(hours)

# Initial temperature setup at midnight (00:00)
T_inside_base[0] = -5.0
T_inside_opt[0] = 15.0

for i in range(1, len(hours)):
    # Total solar heat capture rate (Watts)
    Q_solar_gain = Q_solar_flux[i] * A_window * Solar_SHGC

    # --- CALCULATE BASELINE SHELTER RESPONSE ---
    # Heat transfer rate across structural skin
    Q_loss_base = (T_ambient[i] - T_inside_base[i-1]) / R_base
    # Step change in temperature over delta time (dt)
    dT_base = ((Q_loss_base + Q_solar_gain) / C_base) * dt
    T_inside_base[i] = T_inside_base[i-1] + dT_base

    # --- CALCULATE OPTIMIZED PASSIVE SHELTER RESPONSE ---
    # Heat transfer rate across insulated multi-material skin
    Q_loss_opt = (T_ambient[i] - T_inside_opt[i-1]) / R_opt
    # Step change in temperature showing structural thermal lag
    dT_opt = ((Q_loss_opt + Q_solar_gain) / C_opt) * dt
    T_inside_opt[i] = T_inside_opt[i-1] + dT_opt

# =====================================================================
# 5. MATPLOTLIB VISUALIZATION PREPARATION
# =====================================================================
plt.figure(figsize=(12, 6.5))
plt.grid(True, linestyle='--', alpha=0.6)

# Plotting temperature tracks
plt.plot(hours, T_ambient, label='Ambient Temperature (Outdoor Ladakh)', color='#2c3e50', linewidth=2.5, linestyle=':')
plt.plot(hours, T_inside_base, label='Unoptimized Base Shelter (No Thermal Mass)', color='#e74c3c', linewidth=2)
plt.plot(hours, T_inside_opt, label='Optimized Passive Shelter (With Structural Lag)', color='#2ecc71', linewidth=3)

# Formatting axes, labels, and presentation layout
plt.title('24-Hour Passive Thermal Performance Simulation (Ladakh Core Climate)', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Hour of Day (24h Clock)', fontsize=12, fontweight='semibold')
plt.ylabel('Temperature (°C)', fontsize=12, fontweight='semibold')

plt.xlim(0, 24)
plt.xticks(np.arange(0, 25, 2))
plt.axhline(0, color='black', linewidth=0.8, alpha=0.5) # Freeze point marker

# Highlighting the specific "Sunset Plunge" problem area
plt.axvspan(17, 24, color='#ecf0f1', alpha=0.5, label='Post-Sunset High Thermal Loss Window')

plt.legend(loc='lower left', frameon=True, facecolor='white', edgecolor='none', shadow=True, fontsize=10.5)
plt.tight_layout()

# Render output directly under the Colab workspace cell
plt.show()
```
